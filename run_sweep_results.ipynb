```python
import subprocess

baseline_run = "3k_baseline 2026.05.13--12-17-40"
unet_run = "3k_frozen_pretrained_unet_eval_aligned 2026.08.24--04-07-29"
gaussian_blur_run = "3k_gaussian_blur 2026.05.26--18-42-22"
haar_wavelet_run = "3k_haar_wavelet 2026.05.26--12-12-55"

models = [
    baseline_run,
    gaussian_blur_run,
    haar_wavelet_run,
    unet_run,
]

attacks = [
    "mean",
    "telea",
    "navier",
    "shiftmap",
    "diffusion"
]
```

```python
def run_sweep(run_name, attack, debug=False):
    cmd = [
        "python",
        "sweep_attack_metrics.py",
        "-d", "images",
        "-r", "./runs",
        "--run-name", run_name,
        "--attack", attack,
        "--min-ratio", "0.1",
        "--max-ratio", "0.9",
        "--steps", "9",
        "--seed", "42",
        "-b", "1",
    ]

    if debug:
        cmd.append("--debug")

    print("Running:")
    print(" ".join(cmd))
    print()

    subprocess.run(
        cmd,
        check=True,
    )
```

```python
for run_name in models:
    run_sweep(
        run_name,
        "identity",
        debug=False,
    )
```

```python
# fixed unet-mask, make it align to eval:
for attack in attacks:
    run_sweep(
            unet_run,
            attack,
            debug=True,
        )
```

```python
for attack in attacks:
    run_sweep(
            gaussian_blur_run,
            attack,
            debug=True,
        )
```
